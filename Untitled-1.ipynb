```python
import matplotlib.pyplot as plt

# Jak narysowac od 1 do 100?
# Chcemy mieć wykres od -100 do 100
x = list(range(-100,101))
y = []
z = []

for i in range(-100,101):
    y.append(i ** 2)

for i in range(-100,101):
    z.append(i ** 2 +1000)

plt.plot(x, y,color='green')
plt.plot(x, z,color='pink')
plt.xlabel("Liczby x")
plt.ylabel("Potęgi liczby x")
plt.title("Funkcja kwadratowa")
plt.show()
```

```python
a = 10 
b = "10"

try:
    print(a / b)
except ZeroDivisionError:
    print("Nie dziel przez zero")
except TypeError:
    print("Podaj liczby")
except:
    print("Wystąpił nieoczekiwany błąd")
```

```python
a = 10 
b = 2

try:
    print( a / b)
except ZeroDivisionError:
    print("Nie dziel przez zero")
except TypeError:
    print("Podaj liczby")
except:
    print("Wystąpił nieoczekiwany błąd")
finally:
    print("Koniec programu")
```

```python
def kalk(l:float, l2:float):
    try:
        return l / l2
    except ZeroDivisionError:
        return "Nie dziel przez zero"
    except ValueError: 
        return "Błędna wartość"
        
    finally:
        print("Koniec działania funkcji")

l=10
l2=5

print(kalk(l, l2))
```
